```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
 
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
 
try:
    from IPython.display import display
except ImportError:
    def display(obj):
        print(obj)
 
CSV_PATH = "expense_data_1.csv"
DAYFIRST = False        
FORECAST_DAYS = 30
 
df = pd.read_csv(CSV_PATH)
 
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
display(df.head())
 
print("\nColumn Names:")
print(df.columns.tolist())
 
print("\nDataset Information:")
df.info()
 
print("\nMissing Values:")
print(df.isnull().sum())
 
print("\nIncome and Expense Counts:")
print(df["Income/Expense"].value_counts())
 
rows_before = len(df)
 
df["Date"] = pd.to_datetime(df["Date"], errors="coerce", dayfirst=DAYFIRST)
df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
 
bad_dates = df["Date"].isna().sum()
bad_amounts = df["Amount"].isna().sum()
if bad_dates or bad_amounts:
    print(f"\nWarning: {bad_dates} unparseable dates, "
          f"{bad_amounts} unparseable amounts will be dropped. "
          f"Check DAYFIRST if bad_dates is large.")
 
df = df.dropna(subset=["Date", "Amount", "Category", "Income/Expense"])
df = df[df["Amount"] >= 0].copy()
df = df.sort_values("Date").reset_index(drop=True)
 
print("\nDataset cleaned successfully!")
print(f"Rows before: {rows_before}, after: {len(df)}")
display(df.head())

expenses = df[df["Income/Expense"].str.strip().str.lower() == "expense"].copy()
 
print("Total expense transactions:", len(expenses))
print("\nExpense Categories:")
print(expenses["Category"].value_counts())
display(expenses.head())
 
category_expenses = (
    expenses.groupby("Category")["Amount"].sum().sort_values(ascending=False)
)
 
print("Total Expenses by Category:")
display(category_expenses)
 
plt.figure(figsize=(10, 6))
category_expenses.plot(kind="bar")
plt.title("Total Expenses by Category")
plt.xlabel("Category")
plt.ylabel("Total Expense (₹)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
 
expenses["Month"] = expenses["Date"].dt.to_period("M")
monthly_expenses = expenses.groupby("Month")["Amount"].sum()
 
print("Monthly Expenses:")
display(monthly_expenses)
 
plt.figure(figsize=(10, 5))
monthly_expenses.plot(kind="line", marker="o")
plt.title("Monthly Expense Trend")
plt.xlabel("Month")
plt.ylabel("Total Expenses (₹)")
plt.grid(True)
plt.tight_layout()
plt.show()
 
daily_expenses = expenses.groupby(expenses["Date"].dt.normalize())["Amount"].sum()
daily_expenses.index = pd.to_datetime(daily_expenses.index)
daily_expenses = daily_expenses.sort_index()
 
date_range = pd.date_range(
    start=daily_expenses.index.min(),
    end=daily_expenses.index.max(),
    freq="D",
)
daily_expenses = daily_expenses.reindex(date_range, fill_value=0.0)
daily_expenses.index.name = "Date"
 
daily_df = daily_expenses.reset_index()
daily_df.columns = ["Date", "Daily_Expense"]
 
print("Daily expense data prepared successfully!")
display(daily_df.head(10))
print("Number of days:", len(daily_df))

FEATURES = [
    "Day",
    "DayOfWeek",
    "Previous_Day",
    "Previous_7_Days",
    "Average_Last_7_Days",
]
 
 
def make_features(date, history):
    """Build one feature row for `date` given the list of past daily values."""
    return {
        "Day": date.day,
        "DayOfWeek": date.dayofweek,
        "Previous_Day": history[-1],
        "Previous_7_Days": history[-7],
        "Average_Last_7_Days": float(np.mean(history[-7:])),
    }
 
 
values = daily_df["Daily_Expense"].astype(float).tolist()
rows = []
for i in range(7, len(daily_df)):
    row = make_features(daily_df["Date"].iloc[i], values[:i])
    row["Daily_Expense"] = values[i]
    row["Date"] = daily_df["Date"].iloc[i]
    rows.append(row)
 
feat_df = pd.DataFrame(rows)
 
if len(feat_df) < 30:
    raise ValueError(
        f"Only {len(feat_df)} usable days after feature creation; "
        "need more data for a meaningful train/test split."
    )
 
X = feat_df[FEATURES]
y = feat_df["Daily_Expense"]
 
print("Features created successfully!")
print("\nFeatures used for training:", FEATURES)
display(feat_df.head())

split_index = int(len(feat_df) * 0.80)
 
X_train, X_test = X.iloc[:split_index], X.iloc[split_index:]
y_train, y_test = y.iloc[:split_index], y.iloc[split_index:]
 
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
 
def build_model(regressor):
    prep = ColumnTransformer(
        [("dow", OneHotEncoder(handle_unknown="ignore"), ["DayOfWeek"])],
        remainder="passthrough",
    )
    return Pipeline([("prep", prep), ("model", regressor)])
 
 
linear_model = build_model(LinearRegression())
linear_model.fit(X_train, y_train)
linear_predictions = np.maximum(linear_model.predict(X_test), 0)
print("Linear Regression model trained successfully!")
 
tree_model = build_model(DecisionTreeRegressor(max_depth=4, random_state=42))
tree_model.fit(X_train, y_train)
tree_predictions = np.maximum(tree_model.predict(X_test), 0)
print("Decision Tree model trained successfully!")
 
baseline_predictions = X_test["Average_Last_7_Days"].values
 
def evaluate(name, preds):
    return {
        "Model": name,
        "MAE": mean_absolute_error(y_test, preds),
        "RMSE": np.sqrt(mean_squared_error(y_test, preds)),
        "R2 Score": r2_score(y_test, preds),
    }
 
 
results = pd.DataFrame([
    evaluate("Baseline (7-day avg)", baseline_predictions),
    evaluate("Linear Regression", linear_predictions),
    evaluate("Decision Tree", tree_predictions),
])
 
print("\nModel Evaluation (a model should beat the baseline to be useful):")
display(results.round(2))
 
comparison = pd.DataFrame({
    "Date": feat_df["Date"].iloc[split_index:].values,
    "Actual Expense": y_test.values,
    "Baseline": baseline_predictions,
    "Linear Regression": linear_predictions,
    "Decision Tree": tree_predictions,
})
display(comparison.round(2))
 
plt.figure(figsize=(12, 6))
plt.plot(comparison["Date"], comparison["Actual Expense"], marker="o", label="Actual Expense")
plt.plot(comparison["Date"], comparison["Linear Regression"], marker=".", label="Linear Regression")
plt.plot(comparison["Date"], comparison["Decision Tree"], marker=".", label="Decision Tree")
plt.plot(comparison["Date"], comparison["Baseline"], linestyle="--", label="Baseline")
plt.title("Actual vs Predicted Daily Expenses")
plt.xlabel("Date")
plt.ylabel("Expense (₹)")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
 
history = daily_expenses.astype(float).tolist()
last_date = daily_expenses.index.max()
 
future_dates, future_predictions = [], []
 
for i in range(1, FORECAST_DAYS + 1):
    next_date = last_date + pd.Timedelta(days=i)
    new_data = pd.DataFrame([make_features(next_date, history)])[FEATURES]
 
    prediction = max(0.0, float(linear_model.predict(new_data)[0]))
 
    future_dates.append(next_date)
    future_predictions.append(prediction)
    history.append(prediction)
 
forecast_df = pd.DataFrame({
    "Date": future_dates,
    "Predicted_Daily_Expense": future_predictions,
})
 
print(f"Next {FORECAST_DAYS} Days Expense Forecast:")
display(forecast_df.round(2))
print(
    f"Predicted Total Expense for Next {FORECAST_DAYS} Days: ₹",
    round(forecast_df["Predicted_Daily_Expense"].sum(), 2),
)
 
plt.figure(figsize=(12, 6))
plt.plot(daily_expenses.index, daily_expenses.values, label="Historical Expenses")
plt.plot(
    forecast_df["Date"],
    forecast_df["Predicted_Daily_Expense"],
    marker="o",
    label="Predicted Expenses",
)
plt.title("Historical and Forecasted Daily Expenses")
plt.xlabel("Date")
plt.ylabel("Expense (₹)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
 
print("Highest Spending Category:", category_expenses.idxmax())
print("Total Spent in This Category: ₹", round(category_expenses.max(), 2))
print("Total Recorded Expenses: ₹", round(expenses["Amount"].sum(), 2))
 
df.to_csv("cleaned_expense_data.csv", index=False)
comparison.to_csv("model_predictions.csv", index=False)
forecast_df.to_csv("future_expense_forecast.csv", index=False)
results.to_csv("model_evaluation.csv", index=False)
 
print("Project results saved successfully!")
print("1. cleaned_expense_data.csv")
print("2. model_predictions.csv")
print("3. future_expense_forecast.csv")
print("4. model_evaluation.csv")



results = pd.read_csv("model_evaluation.csv")

display(results)

predictions = pd.read_csv("model_predictions.csv")

display(predictions.head(10))

forecast = pd.read_csv("future_expense_forecast.csv")

display(forecast.head(10))

plt.savefig("expense_analysis.png", dpi=300, bbox_inches="tight")
plt.show()


df = pd.read_csv("cleaned_expense_data.csv")
forecast = pd.read_csv("future_expense_forecast.csv")

print("PERSONAL EXPENSE PREDICTOR")
print("-------------------------")

print("Total transactions:", len(df))

print("\nFuture Expense Forecast:")
display(forecast)

print("\nProject files loaded successfully!")

expenses = df[df["Income/Expense"] == "Expense"]

total_expense = expenses["Amount"].sum()

category_totals = expenses.groupby("Category")["Amount"].sum()

highest_category = category_totals.idxmax()
highest_amount = category_totals.max()

print("========== FINAL PROJECT SUMMARY ==========")
print(f"Total Expense: ₹{total_expense:.2f}")
print("Highest Spending Category:", highest_category)
print(f"Amount Spent in This Category: ₹{highest_amount:.2f}")
print("===========================================")
forecast = pd.read_csv("future_expense_forecast.csv")

print("Future Expense Forecast:")
display(forecast.head(10))

forecast.plot(
    x=forecast.columns[0],
    y=forecast.columns[-1],
    kind="line",
    marker="o",
    figsize=(10, 5),
    title="Personal Expense Prediction"
)

plt.ylabel("Predicted Expense (₹)")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.savefig("expense_analysis.png", dpi=300, bbox_inches="tight")
plt.show()

plt.tight_layout()
plt.savefig("expense_category.png", dpi=300, bbox_inches="tight")
plt.show()

```
